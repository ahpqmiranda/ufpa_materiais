```python
import numpy as np
import sympy as sp
import pandas as pd
import scipy as sc
from IPython.display import Image
```

```python
# comando da questão
Image(filename="q1.png", width=400)
```

```python
# dados da questão
d, T, tau, c, J = sp.symbols('d T tau c J')
L = 1.5 # m
```

```python
# passo 1, identificar as forças
# força 1. peso
rho, V, g  = sp.symbols('rho V g') # rho = densidade, V = volume, g = aceleração da gravidade
g = sc.constants.g
P = rho * V * g # N # m^3
P = P.subs(V, np.pi * (d/2)**2 * L)
P
```

```python
pd.DataFrame({
    'Variável': ['rho (kg/m³)', 'V (m³)', 'g (m/s²)', 'P (N)'],
    'Valor': ['?', f'{np.pi * (d/2)**2 * L}', f'{g}', f'{P}']
})
```

```python
# cálculo da torção
tau_formula = T * c / J
tau_formula = tau_formula.subs({
    c: d/2,
    J: np.pi * (d**4) / 32
})
tau_eq = sp.Eq(tau, tau_formula)
d_eq = sp.solve(tau_eq, d)[0]
d_eq
```

```python
# cálculo da força peso máxima
P_max = sp.symbols('P_max')
P_eq = sp.Eq(P_max, P.subs(d, d_eq))
P_eq
```
