## questão 1:
### de Qual forma o plano de produção (nível estratégico) e o plano mestre de produção (nível tático) podem ser diferenciados?
O plano de produção é um plano de longo prazo, enquanto o plano mestre de produção é um plano de curto prazo.
Vale ressaltar o nível de detalhamento de cada um, o plano de produção é mais abrangente e o plano mestre de produção é mais detalhado.

Horizonte de planejamento:
- Plano de produção: está em semestres ou anos.
- Plano mestre de produção: está em semanas ou meses.

Nível de detalhamento:
- Plano de produção: é mais abrangente, considerando familias de produtos, investimentos, capacidade de produção, etc.
- Plano mestre de produção: é mais detalhado, considerando produtos específicos, quantidades, datas de entrega, etc.

Tomada de decisão:
- Plano de produção: é tomado em nível estratégico, diretores, CEOs e VPs.
- Plano mestre de produção: é tomado em nível tático, gerentes, coordenadores e supervisores.

Relacionamento com a demanda:
- Plano de produção: é mais relacionado com a demanda agregada, tendências de mercado e sazonalidades.
- Plano mestre de produção: é mais relacionado com a demanda específica, pedidos de clientes e previsões de vendas.

# Questão 2:
### Completar os dados do PMP para um determinado produto sendo um lote de 50 unidades e estoque mínimo de 30 unidades.

```python
# Dados:
semanas = [i for i in range(1, 9)]
DP = [330, 330, 335, 335, 350, 350, 350, 350]
DC = [360, 200, 180, 160, 50, 10, 0, 0]
RP = [100, 0, 0, 0, 200, 0, 0, 0]
estoque_inicial = 100
estoque_min = 30
lote = 50
semanas
```

```python

# Cálculo do estoque final:
def estoque_final(estoque_inicial, DP, DC, RP, lote, estoque_min):
    estoque_atual = estoque_inicial
    estoque_final = []
    pmp_semanal = []
    qtd_lotes = []
    for i in range(len(DP)):
        if DP[i] > DC[i]:
            estoque_atual += RP[i] - DP[i]
        else:
            estoque_atual += RP[i] - DC[i]

        if estoque_atual < estoque_min:
            qtd_lote = estoque_min - estoque_atual
            qtd_lote = qtd_lote // lote + 1
            estoque_atual += lote * qtd_lote
            qtd_lotes.append(qtd_lote)
            estoque_final.append(estoque_atual)
            pmp_semanal.append(lote * qtd_lote)

        else:
            estoque_final.append(estoque_atual)
            pmp_semanal.append(0)
            qtd_lotes.append(0)

    return estoque_final, pmp_semanal, qtd_lotes


estoque_final, pmp_semanal, lotes_fab = estoque_final(estoque_inicial, DP, DC, RP, lote, estoque_min)
estoque_final, pmp_semanal, lotes_fab
```

```python
df_prova1 = pd.DataFrame(
    {'Semana': semanas, 'Demanda Planejada': DP, 'Demanda Confirmada': DC, 'Recebimento Planejado': RP,
     'Estoque Final': estoque_final, 'PMP Semanal': pmp_semanal, 'Lotes Fabricados': lotes_fab})
df_prova1.index = df_prova1['Semana']
df_prova1.drop(columns='Semana', inplace=True)
df_prova1.T

```

### Apesar de não agregar valor, os estoques possuem certas funções para o sistema produtivo. Cite pelo menos 4 delas
1. Proteção contra incertezas: os estoques protegem a empresa contra incertezas na demanda, fornecimento e produção.
2. Economia financeira: Os estoques permitem compras em grandes quantidades, o que pode gerar descontos e economias financeiras, além de dar poder de negociação e espera por melhores condições de mercado.
3. Oscilações de produção: Os estoques permitem que a produção seja feita em lotes, o que pode ser mais eficiente e econômico do que produzir sob demanda.
4. Estratégia produtiva: Os estoques podem ser estruturados para produtos semiacabados ou acabados, permitindo uma estratégia de produção mais eficiente e flexível.

# questão 4:
### O que é o lead time e quais são os principais tipos de lead time?
O lead time é o tempo necessário para que um pedido seja entregue ao cliente. Existem vários tipos de lead time, entre eles:
1. Lead time de produção: é o tempo necessário para produzir um item.
2. Lead time de entrega: é o tempo necessário para entregar um item ao cliente.
3. Lead time de reposição: é o tempo necessário para repor um item no estoque.
4. Lead time total: é o tempo total desde o pedido até a entrega ao cliente.
