```python
import numpy as np
import pandas as pd
import scipy as sp
import sympy as sp
```

```python
L_bc = 0.75 #ft
theta_bc = 13.6 # degrees
theta_bc = np.deg2rad(theta_bc) # rad
L_ab = .25 # ft
theta_ab = 45 # degress
theta_ab = np.deg2rad(theta_ab) # rad
phi_ab = 10 # rad/s
```

```python
r_b = np.array([np.cos(theta_ab), np.sin(theta_ab)]) * L_ab
r_b
```

```python
V_b = phi_ab * L_ab
print(f'V_b = {V_b:.2f} ft/s')
```

```python
r_bc = np.array([np.cos(theta_bc), np.sin(theta_bc)]) * L_bc
r_bc
```

```python
V_c = np.array([0, -V_b]) + phi_ab * r_b
print(f'V_c = {V_c}, mag V_c = {np.linalg.norm(V_c):.2f} ft/s')
```
