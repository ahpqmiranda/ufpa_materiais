```python
import numpy as np
import pandas as pd
import scipy as sp
import sympy as sy
```

```python
L_ab = 75  # mm
L_ab = L_ab / 1000  # m
v_rot_ab = 6  # rad/s
L_bc = 100  # mm
L_bc = L_bc / 1000  # m
theta_bc = 30  # graus
theta_bc = np.deg2rad(theta_bc)  # rad
L_cd = 150  # mm
L_cd = L_cd / 1000  # m
theta_cd = 90  # graus
theta_cd = np.deg2rad(theta_cd)  # rad
r_d = 100  # mm
r_d = r_d / 1000  # m
r_f = 25  # mm
r_f = r_f / 1000  # m
```

```python
V_b = v_rot_ab * L_ab
print(f'V_b = {V_b:.2f} m/s')
```

```python
phi_cd, phi_bc = sy.symbols('phi_cd phi_bc')
V_c = phi_cd * L_cd
V_c
```

```python
r_bc = np.array([np.cos(theta_bc), np.sin(theta_bc)]) * L_bc
r_bc
```

```python
V_c = np.array([0, -V_b]) + phi_bc * r_bc
phi_bc = sy.solve(V_c[0], phi_bc)[0]
phi_bc
```

```python
sy.solve(V_c[1], phi_cd)
```
