```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import ace_tools_open as tools
```

```python

# Redefinição das funções para calcular os afastamentos com base na NBR 6158

def afastamento_a(D):
    D_prime = np.sqrt(7 * D)  # Estimativa da média geométrica do intervalo
    if D <= 120:
        return -(265 + 1.3 * D_prime)
    else:
        return -3.5 * D_prime

def afastamento_b(D):
    D_prime = np.sqrt(7 * D)
    if D <= 160:
        return -140 + 0.85 * D_prime
    else:
        return -1.8 * D_prime

def afastamento_c(D):
    D_prime = np.sqrt(7 * D)
    if D <= 40:
        return -52 * (D_prime ** 0.2)
    else:
        return -(95 + 0.8 * D_prime)

def afastamento_d(D):
    D_prime = np.sqrt(7 * D)
    return -16 * (D_prime ** 0.44)

def afastamento_e(D):
    D_prime = np.sqrt(7 * D)
    return -11 * (D_prime ** 0.41)

def afastamento_f(D):
    D_prime = np.sqrt(7 * D)
    return -5.5 * (D_prime ** 0.41)

def afastamento_g(D):
    D_prime = np.sqrt(7 * D)
    return -2.5 * (D_prime ** 0.34)

def afastamento_h(D):
    return 0

# Lista de diâmetros para cálculo
diametros = [40, 150, 270, 25, 110, 40, 15, 63, 85]

# Recalcular os afastamentos para os diâmetros especificados
resultados = []

for D in diametros:
    resultado = {
        "Diâmetro (mm)": D,
        "a_s (a)": afastamento_a(D),
        "a_s (b)": afastamento_b(D),
        "a_s (c)": afastamento_c(D),
        "a_s (d)": afastamento_d(D),
        "a_s (e)": afastamento_e(D),
        "a_s (f)": afastamento_f(D),
        "a_s (g)": afastamento_g(D),
        "a_s (h)": afastamento_h(D)
    }
    resultados.append(resultado)

# Criar dataframe atualizado e exibir
df_resultados_corrigidos = pd.DataFrame(resultados)
tools.display_dataframe_to_user(name="Resultados dos Afastamentos", dataframe=df_resultados_corrigidos)
```
